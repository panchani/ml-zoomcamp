Q7. Sum of weights

```python
# Select all the cars from Asia
cars_from_asia = df[df['origin'] == 'Asia']
cars_from_asia
```

```python
# Select only columns vehicle_weight and model_year
cars_from_asia = cars_from_asia[['vehicle_weight', 'model_year']]
cars_from_asia
```

```python
# Select the first 7 values
cars_from_asia = cars_from_asia.iloc[:7]
cars_from_asia
```

```python
import numpy as np
```

```python
# Get the underlying NumPy array. Let's call it X.
X = cars_from_asia.to_numpy()
X
```

```python
# Compute matrix-matrix multiplication between the transpose of X and X
XT = X.T
XTX = np.dot(XT, X)
XTX
```

```python
# Invert XTX
from numpy.linalg import inv
inverse = inv(XTX)
inverse
```

```python
# Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200]
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
# Multiply the inverse of XTX with the transpose of X, then multiply the result by y
w = np.dot(np.dot(inverse, XT), y)
w
```

```python
# What's the sum of all the elements of the result?
np.sum(w)
```
